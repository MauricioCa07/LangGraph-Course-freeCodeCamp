```python
from typing import TypedDict,Dict
from langgraph.graph import StateGraph
import random
```

```python
class AgentState(TypedDict):
    name : str

def compliment(state: AgentState) -> AgentState:
    """This Function recives the actual state of my graph
       and ads a compliment to the state
    """
    python_compliments = [
    "len(your_flaws) == 0",
    "import happiness as you",
    "my_life.append(you)",
    "while True: love(you)"
    ]

    n = random.randrange(0,3)
    
    state['name'] = "Hey " + state["name"] + " " + python_compliments[n]
    return  state
```

```python
graph = StateGraph(AgentState)

graph.add_node("node1",compliment)


graph.set_entry_point("node1")
graph.set_finish_point("node1")

app = graph.compile()
```

```python
from IPython.display import Image, display
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
result = app.invoke({"name":"bob"})
```
